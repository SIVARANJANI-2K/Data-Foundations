```python
df = df.drop_duplicates()
```

```python
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='%m-%d-%Y %H:%M'
)
```

```python
df[["Order_Date"]]
```

```python
Q1 = df['UnitPrice'].quantile(0.25)
Q3 = df['UnitPrice'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['UnitPrice'] < lower_limit) |
    (df['UnitPrice'] > upper_limit)
]

outliers
```

```python
df['DiscountPct'] = np.where(
    df['Category'] == 'Electronics',
    10,
    5
)

df[['Product', 'Category', 'DiscountPct']]
```

```python
df['Order_Value'] = np.where(
    df['Revenue'] >= 5000,
    'High Value',
    np.where(
        df['Revenue'] >= 1000,
        'Medium Value',
        'Low Value'
    )
)

df[['Order_ID', 'Revenue', 'Order_Value']]
```

```python
df['Year'] = df['Order_Date'].dt.year

df[['Order_ID', 'Order_Date', 'Year']]
```

```python
df['Month'] = df['Order_Date'].dt.to_period('M')

df[['Order_ID', 'Order_Date', 'Month']]
```

```python
monthly_summary = (
    df.groupby('Month')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          Orders=('Order_ID', 'count')
      )
      .sort_index()
)

monthly_summary
```

```python
monthly_summary['TotalRevenue'].idxmax()
```

```python
monthly_summary['TotalRevenue'].idxmin()
```

```python
product_summary = (
    df.groupby('Product')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          UnitsSold=('Quantity', 'sum'),
          Orders=('Order_ID', 'count'),
          AverageRevenue=('Revenue', 'mean')
      )
      .sort_values('TotalRevenue', ascending=False)
)

product_summary
```

```python
category_summary = (
    df.groupby('Category')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          TotalQuantity=('Quantity', 'sum'),
          Orders=('Order_ID', 'count'),
          AverageRevenue=('Revenue', 'mean')
      )
      .sort_values('TotalRevenue', ascending=False)
)

category_summary
```

```python
city_summary = (
    df.groupby('City')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          Orders=('Order_ID', 'count'),
          AverageRevenue=('Revenue', 'mean')
      )
      .sort_values('TotalRevenue', ascending=False)
)

city_summary
```

```python
payment_summary = (
    df.groupby('Payment_Method')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          Orders=('Order_ID', 'count'),
          AverageRevenue=('Revenue', 'mean')
      )
      .sort_values('TotalRevenue', ascending=False)
)

payment_summary
```

```python
df.groupby('Category')['Revenue'].agg(
    ['sum', 'mean', 'count', 'min', 'max']
)
```

```python
region_summary = (
    df.groupby('City')
      .agg(
          TotalRevenue=('Revenue', 'sum'),
          Orders=('Order_ID', 'count'),
          AverageRevenuePerOrder=('Revenue', 'mean')
      )
      .sort_values('TotalRevenue', ascending=False)
)

region_summary
```

```python
df.groupby('Payment_Method')['Revenue'].sum().sort_values(ascending =False)
```

```python
numeric_df= df[['UnitPrice','Quantity','Revenue']]

corr_matrix = numeric_df.corr()
corr_matrix
```

```python
monthly_summary['TotalRevenue'].plot(
    kind='line',
    marker='o',
    title='Monthly Revenue Trends'
)

plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_summary['TotalRevenue'].plot(
    kind='bar',
    title='Total Revenue by City'
)

plt.xlabel('City')
plt.ylabel('Revenue')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
product_summary['TotalRevenue'].sort_values().plot(
    kind='barh',
    title='Total Revenue by Product'
)

plt.xlabel('Revenue')
plt.ylabel('Product')
plt.tight_layout()
plt.show()
```

```python
payment_summary['TotalRevenue'].plot(
    kind='bar',
    title='Revenue by Payment Method'
)

plt.xlabel('Payment Method')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df['UnitPrice'], df['Revenue'])

plt.title('Unit Price vs Revenue')
plt.xlabel('Unit Price')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
plt.hist(df['Revenue'], bins=5)

plt.title('Distribution of Order Revenue')
plt.xlabel('Revenue')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```
