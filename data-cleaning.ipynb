```python
df = df.drop_duplicates()
```

```python
df
```

```python
df["Category"] =df["Category"].str.strip().str.title()
df["Customer_Name"] =df["Customer_Name"].str.strip().str.title()
df["City"] =df["City"].str.strip().str.title()
```

```python
df[["Category","Customer_Name","City"]]
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"],errors="coerce")
```

```python
df[["Order_Date"]]
```

```python
df["UnitPrice"]= pd.to_numeric(df["UnitPrice"],errors="coerce")
df["Quantity"]= pd.to_numeric(df["Quantity"],errors="coerce")
```

```python
df[["UnitPrice","Quantity"]].info()
```

```python
df
```

```python

df['Payment_Method'] = df['Payment_Method'].fillna(
    df['Payment_Method'].mode()[0]
)

df['UnitPrice'] = df['UnitPrice'].fillna(
    df['UnitPrice'].median()
)

df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)

df['Order_Date'] = df['Order_Date'].fillna(
    df['Order_Date'].mode()[0]
)

print(df.isnull().sum())
```

```python
df
```

```python
Q1 = df["UnitPrice"].quantile(0.25)
Q3 = df["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = df[
    (df["UnitPrice"] < lower_bound) |
    (df["UnitPrice"] > upper_bound)
]

outliers[["Order_ID", "Product", "UnitPrice"]]
```

```python
df["UnitPrice"].plot(kind="box", vert=True)

plt.title("Box Plot of Product Prices")
plt.ylabel("UnitPrice")
plt.show()
```

```python
df["Revenue"] = df["UnitPrice"]* df["Quantity"]
```

```python
df[["Order_ID","Product","UnitPrice","Quantity","Revenue"]]
```

```python
def price_categories(price):
    if price<500:
        return "Low"
    elif price<1500:
        return "Medium"
    else:
        return "High"
```

```python
df["Price_category"] = df["UnitPrice"].apply(price_categories)
```
