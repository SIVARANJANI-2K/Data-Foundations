```python
import pandas as pd

df = pd.read_csv("sample_stock_market_data.csv")
```

```python
import numpy as np

difference = np.array(df["High"]) - np.array(df["Low"])

risk = np.where(difference > 150, "High Risk", "Low Risk")

print(risk)
```

```python
risk = []

for i in range(len(df)):
    difference = df["High"].iloc[i] - df["Low"].iloc[i]

    if difference > 150:
        risk.append("High Risk")
    else:
        risk.append("Low Risk")

print(risk)
```

```python
risk = []

for i in range(len(df)):
    difference = df["High"].iloc[i] - df["Low"].iloc[i]

    if difference > 150:
        risk.append("High Risk")
    else:
        risk.append("Low Risk")


df["Risk"] = risk

print(df)
```

```python
df.to_csv("stock_data_with_risk.csv", index=False)

print("CSV file saved successfully!")
```

```python


df["Risk"] = np.where(
    (df["High"] - df["Low"]) > 150,
    "High Risk",
    "Low Risk"
)


df.to_csv("stock_data_with_risk_numpy.csv", index=False)

print(df)
print("CSV file saved successfully!")
```
